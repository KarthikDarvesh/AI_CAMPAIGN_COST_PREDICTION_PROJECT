```python
def plot_scatter_chart(df,brand_name):
  Dedicated = df[(df.brand_name==brand_name) & (df.video_type=='Dedicated')]
  Integrated = df[(df.brand_name==brand_name) & (df.video_type=='Integrated')]
  Conceptual  = df[(df.brand_name==brand_name) & (df.video_type=='Conceptual')]
  Youtube_Short  = df[(df.brand_name==brand_name) & (df.video_type=='Youtube Short')]
  Youtube_Live  = df[(df.brand_name==brand_name) & (df.video_type=='Youtube Live')]

  # plt.figure(figsize=(50,50))
  fig, ax = plt.subplots(1,5, figsize=(40,10))
  merged_all_category = [Dedicated.video_cost, Integrated.video_cost, Conceptual.video_cost, Youtube_Short.video_cost, Youtube_Live.video_cost]

  x = []
  y = []

  for i in merged_all_category:
    Q1 = i.quantile(0.25)
    Q3 = i.quantile(0.75)
    IRQ = Q3-Q1
    lower_limit = Q1-1.5*IRQ
    upper_limit = Q3+1.5*IRQ
    # print(lower_limit, upper_limit)
    x.append(lower_limit)
    y.append(upper_limit)
  print(x)
  print(y)


  custom_label_ax_0 = f'"Dedicated"\n{x[0]:.2f},{y[0]:.2f}'
  custom_label_ax_1 = f'"Integrated"\n{x[1]:.2f},{y[1]:.2f}'
  custom_label_ax_2 = f'"Conceptual"\n{x[2]:.2f},{y[2]:.2f}'
  custom_label_ax_3 = f'"Youtube_Short"\n{x[3]:.2f},{y[3]:.2f}'
  custom_label_ax_4 = f'"Youtube_Live"\n{x[4]:.2f},{y[4]:.2f}'



  # label = ['Dedicated', 'Integrated', 'Conceptual', 'Youtube Short', 'Youtube Live']
  ax[0].boxplot(merged_all_category[0],  patch_artist=True)
  ax[0].set_ylabel("video_cost")
  ax[0].set_xlabel(custom_label_ax_0)
  ax[1].boxplot(merged_all_category[1],  patch_artist=True)
  ax[1].set_ylabel("video_cost")
  ax[1].set_xlabel(custom_label_ax_1)
  ax[2].boxplot(merged_all_category[2],  patch_artist=True)
  ax[2].set_ylabel("video_cost")
  ax[2].set_xlabel(custom_label_ax_2)
  ax[3].boxplot(merged_all_category[3],  patch_artist=True)
  ax[3].set_ylabel("video_cost")
  ax[3].set_xlabel(custom_label_ax_3)
  ax[4].boxplot(merged_all_category[4],  patch_artist=True)
  ax[4].set_ylabel("video_cost")
  ax[4].set_xlabel(custom_label_ax_4)


plot_scatter_chart(data, "INDmoney")
```
